```haskell
sumaLista = foldr (+) 0
mayoresQue10 xs = length (filter (> 10) xs)
productoLista = foldr (*) 1
celsiusAFahrenheit = map (\c -> c * 9 / 5 + 32)
ajustarImpares = map (\x -> if odd x then x + 5 else x)


sumaLista [1,2,3]      
mayoresQue10 [5,11,20,3]
productoLista [2,3,4]    
celsiusAFahrenheit [0,100] 
ajustarImpares [1,2,3,4]   
```

```haskell
nombresLargos = filter (\n -> length n > 5)
soloNegativos = filter (< 0)
edadesUniversitarios = filter (\e -> e >= 18 && e <= 25)

nombresLargos ["Ana","Alejandro","Maria","Fernanda"]
soloNegativos [3,-1,0,-7]                            
edadesUniversitarios [17,18,22,25,26]                                        
```

```haskell
agregarIVA = map (\p -> if p >= 1000 then p * 1.19 else p)
longitudesVocales = map length . filter (\w -> head w `elem` "aeiouAEIOU")
cuadradosPares = map (\x -> x * x) . filter even


agregarIVA [500, 1000, 2000]                   
longitudesVocales ["arbol","casa","Elena","sol"]
cuadradosPares [1,2,3,4]                       
```

```haskell
hayNegativo = any (< 0)
todosPares = all even
import Data.Char (isUpper)

nombresCorrectos = all (\n -> isUpper (head n))

hayNegativo [3, -1, 5]              
hayNegativo [3, 1, 5]               
todosPares [2, 4, 6]                
todosPares [2, 3, 6]               
nombresCorrectos ["Ana", "Luis"]    
nombresCorrectos ["Ana", "luis"]     
```

```haskell
cuadrados = [ x * x | x <- [1..10] ]
divisiblesPor3 = [ x | x <- [1..30], x `mod` 3 == 0 ]
paresOrdenados = [ (x, y) | x <- [1..5], y <- [1..5], x < y ]

cuadrados       
divisiblesPor3  
paresOrdenados   
```

```haskell
divisores :: Int -> [Int]
divisores n = [ d | d <- [1..n], n `mod` d == 0 ]

divisores 12  
divisores 7    
divisores 1    
```

```haskell
divisoresPropios :: Int -> [Int]
divisoresPropios n = [ d | d <- [1..n-1], n `mod` d == 0 ]

esPerfecto :: Int -> Bool
esPerfecto n = sum (divisoresPropios n) == n

divisoresPropios 6   
divisoresPropios 12  
esPerfecto 6         
esPerfecto 28     
esPerfecto 12        
```
